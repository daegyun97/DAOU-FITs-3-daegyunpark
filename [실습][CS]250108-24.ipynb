```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(50)
```

```python
n = 1000
```

```python
data = {
    'customer_id': np.arange(1, n+1),
    'age' : np.random.randint(18, 70, n),
    'tenure' : np.random.randint(1, 10, n),
    'balance' : np.random.randint(0,25000, size =n),
    'num_product' : np.random.randint(1,4,size=n),
    'has_credit_card': np.random.choice([0,1], size = n),
    'is_active_member' : np.random.choice([0,1], size = n),
    'estimated_salary': np.random.randint(30000,150000, size= n),
    'churn' : np.random.choice([0,1], size = n, p=[0.8, 0.2])
    }

df = pd.DataFrame(data)
df.head()
```

```python
df.to_csv('customer_churn.csv', index = False)
```
